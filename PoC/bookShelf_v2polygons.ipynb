```python
def sort_detections(
    detections
):

    if not detections:
        return []

    items = []

    for item in detections:

        x1, y1, x2, y2 = (
            item["box"]
        )

        items.append({

            **item,

            "center_x":
                (x1 + x2) / 2,

            "center_y":
                (y1 + y2) / 2,

            "height":
                y2 - y1
        })

    median_height = np.median(
        [
            item["height"]
            for item in items
        ]
    )

    row_threshold = (
        median_height * 0.5
    )

    items.sort(
        key=lambda x:
            x["center_y"]
    )

    rows = []

    for item in items:

        assigned = False

        for row in rows:

            row_y = np.mean(
                [
                    x["center_y"]
                    for x in row
                ]
            )

            if abs(
                item["center_y"]
                - row_y
            ) <= row_threshold:

                row.append(
                    item
                )

                assigned = True

                break

        if not assigned:

            rows.append(
                [item]
            )

    for row in rows:

        row.sort(
            key=lambda x:
                x["center_x"]
        )

    rows.sort(
        key=lambda row:
            np.mean(
                [
                    x["center_y"]
                    for x in row
                ]
            )
    )

    ordered = []

    book_number = 1

    for row_number, row in enumerate(
        rows,
        start=1
    ):

        for item in row:

            item["id"] = (
                book_number
            )

            item["row"] = (
                row_number
            )

            ordered.append(
                item
            )

            book_number += 1

    return ordered


detections = sort_detections(
    detections
)

print(
    "Ordered detections:",
    len(detections)
)
```

```python
def create_polygon_crop(
    image,
    polygon,
    padding=2
):
    """
    Create two versions of a polygon crop:

    1. white_crop:
       Rectangular image with everything
       outside the polygon turned white.
       Use this for OCR.

    2. rgba_crop:
       Same crop, but outside polygon is
       transparent.
       Use this for visualization/saving.
    """

    polygon = np.asarray(
        polygon,
        dtype=np.float32
    )

    # ------------------------------------
    # Polygon bounding rectangle
    # ------------------------------------

    x, y, w, h = cv2.boundingRect(
        polygon.astype(np.int32)
    )

    # Add small padding.
    x1 = max(
        0,
        x - padding
    )

    y1 = max(
        0,
        y - padding
    )

    x2 = min(
        image.shape[1],
        x + w + padding
    )

    y2 = min(
        image.shape[0],
        y + h + padding
    )

    # ------------------------------------
    # Crop original image
    # ------------------------------------

    crop = image[
        y1:y2,
        x1:x2
    ].copy()

    # ------------------------------------
    # Convert polygon to local coordinates
    # ------------------------------------

    local_polygon = (
        polygon
        - np.array(
            [x1, y1],
            dtype=np.float32
        )
    )

    local_polygon = (
        local_polygon
        .astype(np.int32)
        .reshape(-1, 1, 2)
    )

    # ------------------------------------
    # Create binary mask
    # ------------------------------------

    mask = np.zeros(
        crop.shape[:2],
        dtype=np.uint8
    )

    cv2.fillPoly(
        mask,
        [local_polygon],
        255
    )

    # ------------------------------------
    # White background crop
    # ------------------------------------

    white_crop = np.full_like(
        crop,
        255
    )

    white_crop[
        mask == 255
    ] = crop[
        mask == 255
    ]

    # ------------------------------------
    # Transparent RGBA crop
    # ------------------------------------

    rgba_crop = cv2.cvtColor(
        crop,
        cv2.COLOR_BGR2BGRA
    )

    rgba_crop[:, :, 3] = mask

    return {
        "white_crop":
            white_crop,

        "rgba_crop":
            rgba_crop,

        "mask":
            mask,

        "box":
            (
                x1,
                y1,
                x2,
                y2
            )
    }
```

```python
def upscale_crop(
    crop,
    target_height=1000
):

    if crop is None:
        return crop

    h, w = crop.shape[:2]

    if h == 0 or w == 0:
        return crop

    if h >= target_height:
        return crop

    scale = (
        target_height / h
    )

    new_width = int(
        w * scale
    )

    return cv2.resize(
        crop,
        (
            new_width,
            target_height
        ),
        interpolation=cv2.INTER_CUBIC
    )
```

```python
def parse_ocr_result(
    result
):

    data = result.json

    if isinstance(
        data,
        str
    ):

        data = json.loads(
            data
        )

    if (
        isinstance(data, dict)
        and "res" in data
    ):

        data = data["res"]

    texts = data.get(
        "rec_texts",
        []
    )

    scores = data.get(
        "rec_scores",
        []
    )

    clean_texts = []

    clean_scores = []

    for text, score in zip(
        texts,
        scores
    ):

        text = str(
            text
        ).strip()

        if text:

            clean_texts.append(
                text
            )

            clean_scores.append(
                float(score)
            )

    if not clean_texts:

        return {
            "text": "",
            "confidence": 0.0,
            "lines": []
        }

    combined_text = " ".join(
        clean_texts
    )

    confidence = float(
        np.mean(
            clean_scores
        )
    )

    return {

        "text":
            combined_text,

        "confidence":
            confidence,

        "lines":
            clean_texts
    }
```

```python
def run_ocr(
    image
):

    try:

        results = ocr.predict(
            input=image
        )

        if not results:

            return {
                "text": "",
                "confidence": 0.0,
                "success": False
            }

        parsed = (
            parse_ocr_result(
                results[0]
            )
        )

        parsed["success"] = True

        return parsed

    except Exception as error:

        return {

            "text":
                "",

            "confidence":
                0.0,

            "success":
                False,

            "error":
                str(error)
        }
```

```python
def ocr_book_spine(
    crop,
    retry_threshold=0.75
):

    crop = upscale_crop(
        crop,
        target_height=1000
    )

    # ---------------------------------
    # First: original
    # ---------------------------------

    best = run_ocr(
        crop
    )

    best["orientation"] = (
        "original"
    )

    if (
        best["success"]
        and best["text"].strip()
        and best["confidence"]
        >= retry_threshold
    ):

        return best

    # ---------------------------------
    # Second: clockwise
    # ---------------------------------

    cw = cv2.rotate(
        crop,
        cv2.ROTATE_90_CLOCKWISE
    )

    result_cw = run_ocr(
        cw
    )

    if (
        result_cw["success"]
        and result_cw["text"].strip()
        and
        result_cw["confidence"]
        > best["confidence"]
    ):

        best = {
            **result_cw,
            "orientation":
                "90_clockwise"
        }

    # ---------------------------------
    # Third: counter-clockwise
    # ---------------------------------

    ccw = cv2.rotate(
        crop,
        cv2.ROTATE_90_COUNTERCLOCKWISE
    )

    result_ccw = run_ocr(
        ccw
    )

    if (
        result_ccw["success"]
        and result_ccw["text"].strip()
        and
        result_ccw["confidence"]
        > best["confidence"]
    ):

        best = {
            **result_ccw,
            "orientation":
                "90_counterclockwise"
        }

    return best
```

```python
from tqdm.auto import tqdm

ocr_outputs = []

for book_number, crop in enumerate(
    tqdm(
        book_crops,
        desc="OCR on polygon-masked books"
    ),
    start=1
):

    result = ocr_book_spine(
        crop
    )

    result["book_number"] = (
        book_number
    )

    ocr_outputs.append(
        result
    )

print(
    "OCR completed for",
    len(ocr_outputs),
    "books."
)
```

```python
def ocr_book_spine(
    crop,
    retry_threshold=0.75
):

    crop = upscale_crop(
        crop,
        target_height=800
    )

    # ---------------------------------
    # 1. Original orientation
    # ---------------------------------

    best = run_ocr(
        crop
    )

    best["orientation"] = (
        "original"
    )

    # If already good, stop.
    if (
        best["success"]
        and best["text"].strip()
        and
        best["confidence"]
        >= retry_threshold
    ):

        return best

    # ---------------------------------
    # 2. Rotate clockwise
    # ---------------------------------

    rotated_cw = cv2.rotate(
        crop,
        cv2.ROTATE_90_CLOCKWISE
    )

    result_cw = run_ocr(
        rotated_cw
    )

    if (
        result_cw["success"]
        and result_cw["text"].strip()
        and
        result_cw["confidence"]
        > best["confidence"]
    ):

        best = {
            **result_cw,
            "orientation":
                "90_clockwise"
        }

    # ---------------------------------
    # 3. Rotate counter-clockwise
    # ---------------------------------

    rotated_ccw = cv2.rotate(
        crop,
        cv2.ROTATE_90_COUNTERCLOCKWISE
    )

    result_ccw = run_ocr(
        rotated_ccw
    )

    if (
        result_ccw["success"]
        and result_ccw["text"].strip()
        and
        result_ccw["confidence"]
        > best["confidence"]
    ):

        best = {
            **result_ccw,
            "orientation":
                "90_counterclockwise"
        }

    return best
```
